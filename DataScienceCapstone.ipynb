6. DATA ANALYSIS & FINDINGS

The daily demand data was analyzed to identify trend, seasonality, and residual patterns. The time-series plot showed an increasing trend in demand with a clear 7-day seasonal pattern.

The Augmented Dickey-Fuller (ADF) test indicated that the original series was non-stationary. Differencing and seasonal differencing were therefore considered before SARIMA modeling.

The SARIMA model was trained using historical demand data and evaluated on the test dataset.

Key Findings:

Demand shows an increasing trend.

A weekly (7-day) seasonal pattern is observed.

The original time series is non-stationary.

SARIMA was used to model the demand series.

The model achieved a MAPE of 0.52% and RMSE of 0.93.
